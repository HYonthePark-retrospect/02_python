## Getter/Setter 메서드
##### 캡슐화된 클래스에서 속성을 제어하기 위한 메서드로써 기본적으로 다음 두 유형이 있다.
1. getter 속성의 값을 읽어내기 위한 메서드
   - : instance private 속성의 값을 (사용자가)얻어가게 하는것

2. setter 속성의 값을 설정하기 위한 메서드
   - : instance private 속성에 값을 대입(세팅)

```python
class MyClass:
    def __init__(self, value): # 생성시점 한번만 호출가능 / 여러값 동시 세팅 가능
        self.__value = value

    # getter
    # 외부에서 private 속성 값을 얻어 가게하는 메서드
    def get_value(self):
        return self.__value

    # setter
    # 외부에서 private 속성 값을 변경(세팅)하는 메서드
    def set_value(self, value):
        self.__value = value

my_cls = MyClass(100)
print('my_cls.get_value() : ', my_cls.get_value())

print('200 세팅')
my_cls.set_value(200)
print('my_cls.get_value() : ', my_cls.get_value())
```

## Python의 `@property`

Python에서는 Java식 `get_value()`, `set_value()` 메서드도 사용할 수 있지만, 더 Python다운 방식으로는 `@property`를 자주 사용한다.

`@property`를 사용하면 메서드로 값을 검증하면서도, 사용하는 쪽에서는 속성처럼 읽고 쓸 수 있다.


```python
class Score:
    def __init__(self, value: int):
        self.__value = value

    #은 사람이 보는 주적 @는 파이썬 인터프리터가 보는 주석
    # 오버로딩(Overroding) == 과적, 한 클래스 내에 같은 이름의 메서드가 여러개 존재하는 기술
    @property
    def value(self): # get_value
        return self.__value

    @value.setter
    def value(self, value): #set_value
        if not 0 <= value <= 100:
            raise ValueError('점수는 0~100 사이만 입력해주세요')

        self.__value = value

score = Score(80)
print('score.value : ', score.value)

# score.value(90) Error
score.value = 90 # 메서드 호출X, 변수에 값 대입하듯이 사용
print('score.value : ', score.value)
```
