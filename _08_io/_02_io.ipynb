## 파일 쓰기

```python
f = open('helloworld2.txt', 'w', encoding='utf-8')
print(f, type(f))

text = """
아, 내일이 화요일이라니
나에게 왜 그러는걸까
집에 또 언제가지
"""

f.write(text) # 지정된 파일에 text 출력
# 기존 내용이 있으면 덮어쓰기
# 지정된 파일이 존재하지 않으면 파일 생성

# 파일 연결에 사용된 메모리 반환
f.close()
```

```python
# with 절
# - with 옆에 작성된 open() 구문에서 사용한 파일 연결을
#   with절 종료 시 자동으로 close()
with open('helloworld2.txt', 'w', encoding='utf-8') as f: # as f == f로 별칭을 지정하겠다
    f.write('with 절을 이용하면 자동 close() 된다')
```

## 파일 읽기

```python
with open('helloworld.txt', 'r', encoding='utf-8') as f: # 안에 한글 형식이 있을 수 있기에 utf8을 사용하겠다
    print(f.read(50)) # 전부 읽어와서 출력
```

```python
# 한 줄씩 읽기
with open('helloworld.txt', 'r', encoding='utf-8') as f:
    for i, j in enumerate(f):
        print(i + 1, j, end='') # end=' ' == print() 완료 후 마지막 문자 빈칸
```

```python
# 파일 이어쓰기('a', append 모드)
with open('helloworld.txt', 'a', encoding='utf-8') as f:
    f.write(''' ㅋㅋㅋㅋ 아 졸리다
    ''')

with open('helloworld.txt', 'r', encoding='utf-8') as f:
    print(f.read()) # 이어쓰기 확인
```

```python
# 파일 복사
# 1. 파일을 읽기
with open('helloworld.txt', 'r', encoding='utf-8') as f:
    text = f.read() # with == if or while -> with절에 선언한 변수는 지역변수X -> 전역 변수로 사용 가능

# 2. 파일 쓰기
with open('helloworld_copy.txt', 'w', encoding='utf-8') as f:
    f.write(text)
    print('파일 복사 완료')
```

```python
# 이미지 파일 복사(아무 파일이나 가능)
# - r, w, a == 문자 읽기/쓰기/이어쓰기 모드
# - rb, wb == 바이트 단위 읽기/쓰기 모드
with open('12345.jpg', 'rb') as f: # r, w, a == 문자를 읽기/쓰기/이어쓰기 하는것, 문자를 전제로 사용
    # print(f.read())
    image_bytes = f.read()


with open('copy_image.jpeg', 'wb') as f:
    f.write(image_bytes)
```
